# 02 — Preprocessing

Clean, impute, encode, and save processed OASIS datasets.

**Requirements:** FR-06 through FR-09 (PRD)

```python
import sys
from pathlib import Path

import pandas as pd

sys.path.insert(0, str(Path('..').resolve()))
from src.data_loader import load_and_merge
from src.preprocessing import preprocess_pipeline
from src.utils import data_path, set_seed

set_seed()
PROCESSED_DIR = data_path('processed')
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# Load raw data
df1, df2, merged = load_and_merge()
print(f'Raw shapes — OASIS-1: {df1.shape}, OASIS-2: {df2.shape}')
```

```python
# FR-06 to FR-09: Apply preprocessing pipeline
oasis1_clean = preprocess_pipeline(df1)
oasis2_clean = preprocess_pipeline(df2)
merged_clean = preprocess_pipeline(merged)

print(f'Cleaned shapes — OASIS-1: {oasis1_clean.shape}, OASIS-2: {oasis2_clean.shape}, Merged: {merged_clean.shape}')
print(f'Target distribution:\n{merged_clean["target"].value_counts()}')
```

```python
# Save processed datasets
oasis1_clean.to_csv(PROCESSED_DIR / 'oasis1_cleaned.csv', index=False)
oasis2_clean.to_csv(PROCESSED_DIR / 'oasis2_cleaned.csv', index=False)
merged_clean.to_csv(PROCESSED_DIR / 'oasis_merged_final.csv', index=False)
print('Saved processed CSVs to data/processed/')
```
